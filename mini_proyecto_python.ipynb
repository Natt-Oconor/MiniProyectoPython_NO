## MÓDULO 2: Funciones (Con y Sin Parámetros)

Las funciones nos permiten dividir el código en bloques lógicos y reutilizables.

```python
# A) FUNCIÓN SIN PARÁMETROS: Imprime la portada del sistema
def mostrar_banner():
    print('=' * 70)
    print('       🏬 SISTEMA DE GESTIÓN DE INVENTARIO - TECHSTORE')
    print('=' * 70)

# B) FUNCIÓN CON PARÁMETROS: Calcula el precio con descuento opcional (parámetro por defecto)
def calcular_descuento(precio_base, porcentaje_descuento=10):
    """
    Recibe el precio base y aplica un porcentaje de descuento.
    Si no se pasa el porcentaje, aplica el 10% por defecto.
    """
    descuento = precio_base * (porcentaje_descuento / 100)
    precio_final = precio_base - descuento
    return precio_final, descuento

# Probar las funciones
mostrar_banner()
precio_original = 500.0
precio_con_desc, monto_desc = calcular_descuento(precio_original, porcentaje_descuento=15)
print(f'Precio original: {formatear_moneda(precio_original)}')
print(f'Descuento (15%): -{formatear_moneda(monto_desc)}')
print(f'Precio a pagar: {formatear_moneda(precio_con_desc)}')
```

##  MÓDULO 4: POO - Encapsulamiento

El **encapsulamiento** consiste en ocultar o proteger el estado interno de un objeto para evitar modificaciones no autorizadas o inválidas.

- Los atributos privados se definen con doble guion bajo (`__precio_base`, `__stock`).
- Usamos los decoradores `@property` (getter) y `@atributo.setter` (setter) para validar los cambios.

```python
class Producto:
    """
    Clase Base que representa un producto del inventario.
    Aplica Encapsulamiento en precio y stock.
    """
    def __init__(self, codigo, nombre, precio_base, stock):
        self.codigo = codigo                # Atributo público
        self.nombre = nombre                # Atributo público
        self.__precio_base = 0.0            # Atributo privado
        self.__stock = 0                    # Atributo privado
        
        # Usamos los setters para asignar valores con validación previa
        self.precio_base = precio_base
        self.stock = stock

    # --- ENCAPSULAMIENTO: Getter y Setter de PRECIO --- 
    @property
    def precio_base(self):
        return self.__precio_base

    @precio_base.setter
    def precio_base(self, valor):
        if valor < 0:
            raise ValueError('El precio no puede ser un número negativo.')
        self.__precio_base = valor

    # --- ENCAPSULAMIENTO: Getter y Setter de STOCK --- 
    @property
    def stock(self):
        return self.__stock

    @stock.setter
    def stock(self, cantidad):
        if cantidad < 0:
            raise ValueError('El stock no puede ser negativo.')
        self.__stock = cantidad

    # Método general de información
    def obtener_resumen(self):
        return f"[{self.codigo}] {self.nombre} - Precio Base: {formatear_moneda(self.precio_base)} | Stock: {self.stock}"

    # Método que será sobrescrito (Polimorfismo)
    def calcular_precio_final(self):
        # Añade el IVA base al precio
        return self.precio_base * (1 + mi.IMPUESTO_IVA)

# Probar la clase base
prod_test = Producto('PROD-001', 'Mouse Óptico', 25.0, 10)
print(prod_test.obtener_resumen())
print(f"Precio Final con IVA: {formatear_moneda(prod_test.calcular_precio_final())}")
```

##  MÓDULO 5: POO - Herencia y Polimorfismo

- **Herencia**: Las clases Hijas (`ProductoElectronico` y `ProductoAlimento`) heredan los atributos y métodos de la clase Padre (`Producto`).
- **Polimorfismo**: Cada clase Hija redefine el método `calcular_precio_final()` y `obtener_resumen()` a su propia manera.

```python
# 1. CLASE HIJA 1: ProductoElectrónico (Hereda de Producto)
class ProductoElectronico(Producto):
    def __init__(self, codigo, nombre, precio_base, stock, garantia_meses=12):
        # Llamamos al constructor de la clase Padre
        super().__init__(codigo, nombre, precio_base, stock)
        self.garantia_meses = garantia_meses

    # POLIMORFISMO: Redefinición de calcular_precio_final (Agrega tasa de garantía extendida)
    def calcular_precio_final(self):
        precio_con_iva = super().calcular_precio_final()
        # Si tiene más de 12 meses de garantía, se cobra $15 adicionales por seguro
        costo_garantia = 15.0 if self.garantia_meses > 12 else 0.0
        return precio_con_iva + costo_garantia

    # POLIMORFISMO: Redefinición de obtener_resumen
    def obtener_resumen(self):
        resumen_padre = super().obtener_resumen()
        return f"{resumen_padre} | Garantía: {self.garantia_meses} meses (Electrónico)"

# 2. CLASE HIJA 2: ProductoAlimento (Hereda de Producto)
class ProductoAlimento(Producto):
    def __init__(self, codigo, nombre, precio_base, stock, dias_para_vencer):
        super().__init__(codigo, nombre, precio_base, stock)
        self.dias_para_vencer = dias_para_vencer

    # POLIMORFISMO: Redefinición de calcular_precio_final (Aplica liquidación si vence pronto)
    def calcular_precio_final(self):
        precio_con_iva = super().calcular_precio_final()
        # Si le quedan 3 días o menos para vencer, recibe 40% de descuento automático
        if self.dias_para_vencer <= 3:
            print(f"⚠️ ¡ALERTA LIQUIDACIÓN! {self.nombre} vence en {self.dias_para_vencer} días (40% OFF).")
            return precio_con_iva * 0.60
        return precio_con_iva

    # POLIMORFISMO: Redefinición de obtener_resumen
    def obtener_resumen(self):
        resumen_padre = super().obtener_resumen()
        return f"{resumen_padre} | Vence en: {self.dias_para_vencer} días (Alimento)"

# --- DEMOSTRACIÓN DE POLIMORFISMO ---
# Creamos una lista heterogénea de productos (Objetos de distintas clases)
lista_productos = [
    ProductoElectronico('ELEC-101', 'Monitor 27 FPS', 300.0, 8, garantia_meses=24),
    ProductoAlimento('ALIM-202', 'Café de Grano 1Kg', 15.0, 20, dias_para_vencer=2),
    ProductoElectronico('ELEC-103', 'Teclado Mecánico', 60.0, 12, garantia_meses=12)
]

print('✨ DEMOSTRACIÓN DE POLIMORFISMO (Un mismo método calcular_precio_final() responde distinto):')
print('-' * 85)
for p in lista_productos:
    print(p.obtener_resumen())
    print(f"👉 Precio Final Calculado: {formatear_moneda(p.calcular_precio_final())}\n")
```

##  MÓDULO 6: Manejo de Excepciones y Excepciones Personalizadas

El manejo de excepciones permite capturar errores de ejecución para evitar que el programa colapse inesperadamente.

```python
# 1. Definimos nuestra propia Excepción Personalizada
class StockInsuficienteError(Exception):
    """Excepción lanzada cuando se intenta vender más stock del disponible."""
    def __init__(self, solicitado, disponible):
        super().__init__(f"Stock insuficiente: Solicitado {solicitado}, pero solo hay {disponible} unidades disponibles.")
        self.solicitado = solicitado
        self.disponible = disponible

# 2. Función para realizar una venta con captura de errores
def procesar_venta(producto, cantidad_deseada):
    print(f"\n🛒 Intentando comprar {cantidad_deseada} unidades de '{producto.nombre}'...")
    
    try:
        # Validar tipo de dato de entrada
        if not isinstance(cantidad_deseada, int) or cantidad_deseada <= 0:
            raise TypeError('La cantidad comprada debe ser un número entero positivo mayor a cero.')
            
        # Validar disponibilidad de stock
        if cantidad_deseada > producto.stock:
            raise StockInsuficienteError(cantidad_deseada, producto.stock)
            
        # Si todo es correcto, descontar stock
        producto.stock -= cantidad_deseada
        total_pagar = producto.calcular_precio_final() * cantidad_deseada
        print(f"✅ Venta exitosa. Total cobrado: {formatear_moneda(total_pagar)}")
        print(f"📦 Stock restante de '{producto.nombre}': {producto.stock} unidades.")
        
    except StockInsuficienteError as error_stock:
        print(f"❌ ERROR DE STOCK: {error_stock}")
        
    except TypeError as error_tipo:
        print(f"❌ ERROR DE DATO: {error_tipo}")
        
    except Exception as error_general:
        print(f"❌ ERROR INESPERADO: {error_general}")
        
    finally:
        print("🔹 (Bloque 'finally': Operación finalizada e historial de auditoría actualizado).")

# --- PRUEBAS DE EXCEPCIONES ---
laptop = ProductoElectronico('LAP-001', 'Laptop HP Pro', 800.0, 3)

# Prueba 1: Venta exitosa
procesar_venta(laptop, 2)

# Prueba 2: Excepción por Stock Insuficiente (Queda 1 en stock, solicitamos 5)
procesar_venta(laptop, 5)

# Prueba 3: Excepción por tipo de dato inválido
procesar_venta(laptop, -2)
```

##  MÓDULO 7: Integración del Gestor de Inventario

A continuación, integramos los productos y el catálogo en una clase gestora.

```python
class GestorInventario:
    """
    Clase Gestora que almacena los productos en una lista y diccionario,
    demostrando la interacción de todos los conceptos requeridos.
    """
    def __init__(self, nombre_tienda):
        self.nombre_tienda = nombre_tienda
        self.inventario_lista = []       # Lista de objetos Producto
        self.inventario_dict = {}        # Diccionario con clave = código de producto

    def agregar_producto(self, producto):
        if not isinstance(producto, Producto):
            raise TypeError('Solo se pueden agregar objetos que deriven de la clase Producto.')
        self.inventario_lista.append(producto)
        self.inventario_dict[producto.codigo] = producto
        print(f"➕ Producto registrado: [{producto.codigo}] {producto.nombre}")

    def buscar_por_codigo(self, codigo):
        try:
            return self.inventario_dict[codigo]
        except KeyError:
            print(f"⚠️ Error: El producto con código '{codigo}' no existe en la tienda.")
            return None

    def mostrar_reporte_completo(self):
        mostrar_banner()
        print(f"📍 REPORTE GENERAL DE INVENTARIO | {mi.obtener_timestamp()}")
        print('-' * 85)
        
        total_valor_inventario = 0.0
        for prod in self.inventario_lista:
            # Polimorfismo en acción al llamar a obtener_resumen() y calcular_precio_final()
            print(prod.obtener_resumen())
            precio_f = prod.calcular_precio_final()
            valor_subtotal = precio_f * prod.stock
            total_valor_inventario += valor_subtotal
            print(f"   💰 Precio Final c/Impuesto: {formatear_moneda(precio_f)} | Valor Total Stock: {formatear_moneda(valor_subtotal)}\n")
            
        print('=' * 85)
        print(f"💵 VALOR TOTAL DEL INVENTARIO EN TIENDA: {formatear_moneda(total_valor_inventario)}")
        print('=' * 85)

# === INSTANCIA DE TIENDA CON PRODUCTOS BASE ===
mi_tienda = GestorInventario('TechStore Nicaragua')

# Creación de Productos iniciales usando Módulo Custom
prod1 = ProductoElectronico(generar_codigo_unico('TV'), 'Smart TV 55 Pulgadas', 550.0, 4, garantia_meses=24)
prod2 = ProductoAlimento(generar_codigo_unico('SNK'), 'Chocolates Gourmet', 8.50, 30, dias_para_vencer=1)
prod3 = ProductoElectronico(generar_codigo_unico('AUD'), 'Audífonos Noise Cancelling', 120.0, 10, garantia_meses=12)

mi_tienda.agregar_producto(prod1)
mi_tienda.agregar_producto(prod2)
mi_tienda.agregar_producto(prod3)
```

##  MÓDULO 8: Menú Interactivo con Validación de Entradas

Este módulo implementa el **flujo completo de Entrada, Procesamiento y Salida de información** con control de excepciones y ciclo interactivo.

```python
def menu_interactivo(tienda):
    """
    Función interactiva que valida ENTRADA, PROCESAMIENTO y SALIDA
    de datos mediante un ciclo continuo tolerante a errores.
    """
    while True:
        print("\n" + "=" * 55)
        print(f"   🏬 MENÚ DE CONTROL INTERACTIVO - {tienda.nombre_tienda.upper()}")
        print("=" * 55)
        print("1. ➕ Registrar Producto Electrónico")
        print("2. 🥖 Registrar Producto Alimento")
        print("3. 🛒 Realizar Venta (Validación de Stock y Excepciones)")
        print("4. 📊 Ver Reporte General de Inventario")
        print("5. 🚪 Salir del Sistema")
        print("=" * 55)
        
        opcion = input("Seleccione una opción (1-5): ").strip()
        
        if opcion == "1":
            # ENTRADA DE DATOS CON VALIDACIÓN
            try:
                nombre = input("Nombre del producto electrónico: ").strip()
                if not nombre:
                    raise ValueError("El nombre no puede estar vacío.")
                precio = float(input("Precio base ($): "))
                stock = int(input("Cantidad inicial en stock: "))
                garantia = int(input("Meses de garantía (ej. 12, 24): "))
                
                # PROCESAMIENTO
                codigo = mi.generar_codigo_unico("ELEC")
                nuevo_prod = ProductoElectronico(codigo, nombre, precio, stock, garantia)
                tienda.agregar_producto(nuevo_prod)
                
                # SALIDA
                print(f"✅ ¡Producto registrado con éxito! Código asignado: {codigo}")
            except ValueError as e:
                print(f"❌ ERROR EN ENTRADA: {e}")
                
        elif opcion == "2":
            try:
                nombre = input("Nombre del alimento: ").strip()
                if not nombre:
                    raise ValueError("El nombre no puede estar vacío.")
                precio = float(input("Precio base ($): "))
                stock = int(input("Cantidad inicial en stock: "))
                dias_vence = int(input("Días para que venza (ej. 2, 15): "))
                
                codigo = mi.generar_codigo_unico("ALIM")
                nuevo_prod = ProductoAlimento(codigo, nombre, precio, stock, dias_vence)
                tienda.agregar_producto(nuevo_prod)
                print(f"✅ ¡Alimento registrado con éxito! Código asignado: {codigo}")
            except ValueError as e:
                print(f"❌ ERROR EN ENTRADA: {e}")

        elif opcion == "3":
            codigo = input("Ingrese el código del producto a vender (ej. ELEC-101): ").strip().upper()
            prod = tienda.buscar_por_codigo(codigo)
            
            if prod:
                try:
                    cant = int(input(f"Stock disponible de '{prod.nombre}' ({prod.stock} uds). ¿Cuántas desea comprar?: "))
                    procesar_venta(prod, cant)
                except ValueError:
                    print("❌ ERROR: Debe ingresar un número entero válido.")
                    
        elif opcion == "4":
            tienda.mostrar_reporte_completo()
            
        elif opcion == "5":
            print("\n👋 Gracias por utilizar el Sistema de Gestión TechStore. ¡Hasta pronto!")
            break
        else:
            print("⚠️ Opción inválida. Por favor, ingrese un número del 1 al 5.")

# --- EJECUTAR EL MENÚ INTERACTIVO ---
menu_interactivo(mi_tienda)
```
